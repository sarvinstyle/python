```python
from sqlalchemy import create_engine , Column , Integer , String , ForeignKey
from sqlalchemy.orm import relationship , declarative_base

engine = create_engine("sqlite:///orm.db")

Base = declarative_base()

class User(Base):
    __tablename__ = 'users'

    id = Column(Integer, primary_key=True)
    name = Column(String(100) )
    email = Column(String)

    posts = relationship("Post" , backref="author")


class Post(Base):
    __tablename__ = "posts"

    id = Column(Integer, primary_key=True)
    title = Column(String(250))
    content =Column(String)
    user_id = Column(Integer , ForeignKey("users.id"))



Base.metadata.create_all(engine)
```

```python
from sqlalchemy.orm import sessionmaker

Session = sessionmaker(bind=engine)
session = Session()

user1=User(name = "sarvin" , email="sarvin@some.ir")
user2=User(name = "mina" , email="mina@some.ir")

post1 = Post(title = "title1" , content ="content1" , author = user1)
post2= Post(title = "title2" , content ="content2" , author = user1)
post3= Post(title = "title3" , content ="content3" , author = user2)

session.add_all([user1 , user2 , post1 , post2 , post3])
session.commit()
```

```python
users = session.query(User).all()
for user in users:
    print(user.id , user.name , user.email)

posts = session.query(Post).all()
for p in posts:
    print(p.title , p.content , p.user_id)

sarvin = session.query(User).filter_by(name="sarvin").first()
print(sarvin.name , sarvin.email)

filter_post = session.query(Post).join(User).filter(User.name == "sarvin").all()
for p in filter_post:
    print(p.title , p.content , p.user_id)
```

```python
sarvin = session.query(User).filter_by(name="sarvin").first()
if sarvin:
    sarvin.name = "sarvintaba"
    session.commit()
```
