```python
df = pd.read_csv('./world_population.csv')
df.head(10)
df.tail(15)
df.sample(10)

#df.info()
#df.describe()
df.shape
df.columns
df['Country'].describe()


# subsetting

df[df.columns[-5:]]
df_cr = df[['Country' , 'Rank']]

# loc iloc
df.iloc[5:10]
df.iloc[5:10 , 3 : 9]
df.iloc[[5]]

#loc
df.loc[:, ["Country" ,"Capital" ,"Rank" ]]

df2 = df.loc[(df["Rank"] > 90) & (df["Rank"] < 100) ][["Country" ,"Capital" ,"Rank"]]
df2.reset_index()


df[(df["Rank"] >90) & (df["Rank"] < 100) ][["Country" ,"Capital" ,"Rank"]]
df[df["Country"].str.contains("A" , case= False)]


# query
num = 130
df.query('(Rank > @num) and Country == "Albania"')

df[["2022 Population" , "2020 Population" , "2015 Population"]].agg(['min' ,'max','mean'])
df["Continent"].value_counts()
df["Continent"].unique()

# group by 
df.groupby("Continent")[["Rank"]].agg(['min' , 'max' , 'mean' , 'count'])

#df["new"]= df["Rank"] / 2
#df = df.assign(new2 = df["Rank"] / 4)

df.sort_values("Continent" , ascending=False)
df.sort_values(["Continent" , "Country"], ascending=[1,0])

# missing value
df[["Country", "2020 Population" , "2015 Population"]].isna()
df[["Country", "2020 Population" , "2015 Population"]].dropna()
df[["Country", "2020 Population" , "2015 Population"]].fillna(df["2020 Population"].mean())


# merg
df_asia =(df[df["Continent"]== "Asia"]).copy()
df_africa =(df[df["Continent"]== "Africa"]).copy()

result = pd.concat([df_asia , df_africa] )#, axis= 1
result

df_rate =pd.DataFrame( {"Continent" : ["Asia", "Africa" , "Eroupe"],
        "Rate" : [10 ,5, 1]
        })
df_rate

df_final = pd.merge(df, df_rate , left_on=["Continent"],  right_on=["Continent"])
df_final.head()

df_final.to_("output.csv")
```
